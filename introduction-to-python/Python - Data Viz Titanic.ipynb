# Who are the survivors of the Titanic?

#### Survival rate

```python
df.groupby('Survived')['PassengerId'].agg(['count'])
```

```python
df.groupby('Survived')['PassengerId'].agg(
    ['count']
).reset_index().plot(x='Survived', y='count', kind = 'bar', figsize = (10, 10))
```

```python
piechart = df.groupby('Survived')['PassengerId'].agg(
    ['count'])
piechart.plot.pie(y='count', autopct='%1.1f%%', figsize = (5, 5),colors=['red','blue'],labels=['Died','Survived'],title='Survival Rate')
```

#### Survival rate depending on genre

```python
df.groupby(
    ['Survived', 'Sex']
)['PassengerId'].agg(['count'])
```

```python
df.groupby(
    ['Survived', 'Sex']
)['PassengerId'].agg(['count']).unstack()
```

```python
df.groupby(
    ['Survived', 'Sex']
)['PassengerId'].count().unstack().plot(kind ='barh', figsize = (10, 10))
```

```python
import matplotlib.pyplot as plt

proportions = (
    df.groupby(['Sex', 'Survived'])['PassengerId']
      .count()
      .unstack()
)

proportions = proportions.div(proportions.sum(axis=1), axis=0)

proportions.plot(
    kind='barh',
    stacked=True,
    figsize=(10, 5),
    color=['red', 'blue']
)

plt.xlabel('Proportion')
plt.ylabel('Sexe')
plt.legend(['Décédé', 'Survivant'])
plt.show()
```

```python
df.groupby(
    ['Survived', 'generation']
)['PassengerId'].count().unstack().plot(kind ='bar', figsize = (10, 10))
```

```python
proportions = (
    df.groupby(['generation', 'Survived'])['PassengerId']
      .count()
      .unstack()
)

proportions = proportions.div(
    proportions.sum(axis=1),
    axis=0
)

proportions.plot(
    kind='bar',
    stacked=True,
    figsize=(10, 6)
)

plt.xlabel('Génération / tranche d’âge')
plt.ylabel('Proportion')
plt.title('Survie selon la tranche d’âge')
plt.legend(['Décédé', 'Survivant'])

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.regplot(
    data=df,
    x='Age',
    y='Survived',
    logistic=False,
    scatter_kws={'alpha': 0.3},
    line_kws={'color': 'red'}
)

plt.xlabel('Âge')
plt.ylabel('Survie (0 = mort, 1 = survivant)')
plt.title('Relation entre l’âge et la survie')
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 7))

# Couleurs = sexe
couleurs = {
    'male': '#1976D2',      # bleu
    'female': '#E91E63'     # rose
}

# Style du trait = classe
styles = {
    1: '-',      # continu
    2: '--',     # tirets
    3: ':'       # pointillés
}

for classe in [1, 2, 3]:
    for sexe in ['male', 'female']:

        df_filtre = df[
            (df['Pclass'] == classe) &
            (df['Sex'] == sexe)
        ]

        sns.regplot(
            data=df_filtre,
            x='Age',
            y='Survived',
            logistic=False,
            scatter_kws={
                'alpha': 0.08
            },
            line_kws={
                'color': couleurs[sexe],
                'linestyle': styles[classe],
                'linewidth': 2.5
            }
        )

# Labels
plt.xlabel('Âge')
plt.ylabel('Survie (0 = mort, 1 = survivant)')
plt.title('Relation entre l’âge et la survie selon la classe et le sexe')

plt.ylim(-0.1, 1.1)

# Légende personnalisée
from matplotlib.lines import Line2D

legende = [
    Line2D([0], [0], color='#1976D2', lw=3, label='Homme'),
    Line2D([0], [0], color='#E91E63', lw=3, label='Femme'),
    Line2D([0], [0], color='black', lw=2.5, linestyle='-', label='Classe 1'),
    Line2D([0], [0], color='black', lw=2.5, linestyle='--', label='Classe 2'),
    Line2D([0], [0], color='black', lw=2.5, linestyle=':', label='Classe 3')
]

plt.legend(
    handles=legende,
    title='Sexe et classe',
    bbox_to_anchor=(1.02, 1),
    loc='upper left'
)

plt.grid(alpha=0.2)
plt.tight_layout()

plt.show()
```
